```python
class IsOddEven:
    def __init__(self,number : int ):
        self.number=number 

    def isOdd(self):
        if self.number % 2 != 0:
            return True
        else:
            return False

    def isEven(self):
        if self.number % 2 == 0:
            return True
        else:
            return False
IsOddEven1=IsOddEven(5)
print(IsOddEven1.isOdd())
```
